```python
import re
with open(MED_TRAIN_DATA_PATH) as f:
    data = f.readlines()
    pattern = re.compile(r"^\d+\s+(\S+)\s+\S+\s+\S+\s+(\S+)")
    tokens, labels = [], []
    for i in range(len(data)):
        line = pattern.search(data[i])
        if line:
            tokens.append(line.group(1))
            labels.append(line.group(2))
print(tokens[:10])
print(labels[:10])
```
